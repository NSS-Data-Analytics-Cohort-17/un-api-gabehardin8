In this exercise, you'll redo the data gathering phase of the UNData Exploration project by using APIs instead of downloading csv files.

You'll make use of the [World Bank Indicators API](https://datahelpdesk.worldbank.org/knowledgebase/articles/889392-about-the-indicators-api-documentation). 
Note that this API does not require an API key. Before attempting the exercise, it would be a good idea to skim through the Documentation page and to check out the 
[Basic Call Structure article](https://datahelpdesk.worldbank.org/knowledgebase/articles/898581).

1. Use the API to get all available data for the _GDP per capita, PPP (constant 2017 international $)_ indicator. Hint: this indicator has code "NY.GDP.PCAP.PP.KD".
Adjust the query parameters so that you can retrieve all available rows. Convert the results to a DataFrame.

```python
endpoint = 'https://api.worldbank.org/v2/country/all/indicator/NY.GDP.PCAP.PP.KD'
```

```python
params = {
    'format': 'json',
    'per_page': '20000'
}
response = requests.get(endpoint,params = params)
response
```

```python
res = response.json()
res
```

```python
res[1]
```

```python
gdp_per_capita = pd.json_normalize(res[1])
gdp_per_capita.head()
```

```python
gdp_per_capita.columns
```

```python
gdp_per_capita = gdp_per_capita.drop(columns = ['countryiso3code','unit', 'obs_status', 'decimal','indicator.id', 'indicator.value', 'country.id'])
gdp_per_capita.head()
```

```python
gdp_per_capita = gdp_per_capita.rename(columns = {'date': 'Year', 'value': 'GDP Per Capita', 'country.value':'Country'})
gdp_per_capita.head()
```

2. Now, use the API to get all available data for _Life expectancy at birth, total (years)_. This indicator has code "SP.DYN.LE00.IN". Again, convert the results to a DataFrame.

```python
endpoint = 'https://api.worldbank.org/v2/country/all/indicator/SP.DYN.LE00.IN'
```

```python
params = {
    'format': 'json',
    'per_page': '20000'
}
response = requests.get(endpoint,params = params)
response.json()
```

```python
res = response.json()
```

```python
res
```

```python
life_expectancy = pd.json_normalize(res[1])
life_expectancy.head()
```

```python
life_expectancy = life_expectancy.drop(columns = ['countryiso3code','unit', 'obs_status', 'decimal','indicator.id', 'indicator.value', 'country.id'])
life_expectancy.head()
```

```python
life_expectancy = life_expectancy.rename(columns = {'date': 'Year', 'value': 'Life Expectancy', 'country.value':'Country'})
life_expectancy.head()
```

3. Merge the two results DataFrames together. You may want to rename or drop columns prior to merging.

```python
gdp_life = pd.DataFrame.merge(life_expectancy,gdp_per_capita, how = 'inner', on = ['Year','Country'])
gdp_life.head()
```

```python
gdp_life = gdp_life[['Year', 'Country','Life Expectancy','GDP Per Capita']]
gdp_life.head()
```

4. You can also get more information about the available countries (region, capital city, income level classification, etc.) by using the 
[Country API](https://datahelpdesk.worldbank.org/knowledgebase/articles/898590-country-api-queries). Use this API to pull in all available data. 
Merge this with your other datasets. Use this to now remove the rows that correspond to regions and not countries.

```python
endpoint = 'https://api.worldbank.org/v2/country'
```

```python
params = {
    'format': 'json',
    'per_page': '20000'
}
response = requests.get(endpoint,params = params)
response
```

```python
res = response.json()
res
```

```python
country_classification = pd.json_normalize(res[1])
country_classification.head()
```

```python
country_classification.columns
```

```python
gdp_life.columns
```

```python
country_gdp_life = pd.DataFrame.merge(gdp_life, country_classification, how = 'outer', right_on = 'name', left_on = 'Country') 
country_gdp_life
```

```python
country_gdp_life = country_gdp_life.loc[country_gdp_life['region.id'] != 'NA']
country_gdp_life
```
